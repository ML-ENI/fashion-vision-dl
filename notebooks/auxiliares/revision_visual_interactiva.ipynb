## Qué se guarda y qué queda pendiente

Cada clic actualiza `config/visual_review_pending_chat.csv`. Ya contiene las cuatro decisiones de la revisión del chat: tres exclusiones y los leggings como duda. Al cerrar y volver a abrir, la galería relee ese archivo y continúa con las fotos sin revisar.

Este es un **registro pendiente**: el código actual de 02 todavía no lo aplica. No modifica `training_visual_review.csv`, las fotos, las particiones ni los resultados. Cuando acabes una tanda, podremos incorporar las decisiones confirmadas al registro activo en un único paso. Las dudas quedarán pendientes. Así evitamos repetir la preparación o el entrenamiento por cada clic.

La galería solo lee metadatos y las miniaturas que se muestran; no ejecuta CLIP, entrenamientos ni prepara el dataset. El resumen de decisiones cuenta únicamente el registro pendiente, no una validación completa del dataset.

**Filtro inicial:** solo dudas de CLIP que siguen en el dataset y que tú aún no has revisado. No es todo el conjunto. Las fotos que CLIP consideró buenas quedan fuera de este filtro, aunque también podrían contener errores. «Mis dudas guardadas» muestra tus propias marcas de duda, no las de CLIP.
