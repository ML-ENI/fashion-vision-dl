# 10 — Final Comparison, Test Evaluation and Interpretability

This notebook closes the supervised study. The best configuration of each family — Dense, CNN, LSTM / GRU, Transformer and pretrained model — is evaluated **once** on the untouched test partition, after all design decisions have been taken with validation data.

It compares models on performance, size and cost, analyses per-class behaviour and errors, and uses interpretability tools (e.g. Grad-CAM, saliency maps, attention maps) to check whether predictions rely on the garment itself and not on background or photographic artefacts.

> **Status: skeleton.** Only the introduction, objectives and work preview are written. Code and results will be added when the phase is developed.

## Objectives

- Freeze the final configuration of every architecture family and retrain if required.
- Evaluate each model once on the test split: accuracy, macro F1, per-class precision and recall, confusion matrix.
- Compare parameters, training time, inference time and GPU memory.
- Analyse confusable classes and misclassified examples.
- Apply Grad-CAM / saliency / attention maps to explain predictions.
- Summarise conclusions, limitations and future work for the PRDL and MLLB reports.

## Preview of the work

1. **Final model registry.** checkpoints and configurations selected in notebooks 04–09.
2. **Test evaluation.** single pass on the test split with bootstrapped confidence intervals.
3. **Model comparison.** performance vs. complexity and cost.
4. **Per-class and error analysis.** confusion matrix, hardest classes, typical errors.
5. **Interpretability.** Grad-CAM for CNNs, attention maps for ViT, comparison across models.
6. **Conclusions and limitations.** label noise, class taxonomy, image style, compute budget.
