## Objetivo y relación con los otros bloques

Modelamos `p(x_1,...,x_T) = producto_t p(x_t | x_<t)`. Durante el entrenamiento la entrada es una secuencia desplazada respecto al objetivo: cada posición predice el siguiente carácter. Durante la generación se añade un carácter muestreado y se vuelve a predecir. Una GRU es recurrente; un Transformer decoder requiere impedir el acceso a tokens futuros mediante máscara causal. BERT original usa reconstrucción bidireccional de tokens enmascarados y no es el ejemplo de generación causal elegido.

El objetivo de siguiente token produce señales a partir del propio texto y conecta autosupervisión con generación. No convierte un modelo pequeño entrenado aquí en un modelo fundacional. Embeddings, CLIP zero-shot y clasificación visual no sustituyen la generación.
