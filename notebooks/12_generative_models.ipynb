# 12 — Generative Models: Autoencoders and VAE

This notebook explores **generative models**, which learn the distribution of the training data and can sample new examples from it. A convolutional autoencoder and a **variational autoencoder (VAE)** are trained on H&M garment images to learn a compact latent representation without using the labels.

The latent space is analysed (structure by class, interpolation between garments, generation of new samples), and the learned encoder is reused as a feature extractor for classification, linking generative modelling with self-supervised representation learning.

> **Status: skeleton.** Only the introduction, objectives and work preview are written. Code and results will be added when the phase is developed.

## Objectives

- Train a convolutional autoencoder and a VAE on low-resolution garment images.
- Explain the VAE objective: reconstruction term plus KL divergence and the reparameterisation trick.
- Visualise reconstructions, random samples and latent-space interpolations.
- Project the latent space in 2-D coloured by garment class.
- Use the encoder as a frozen feature extractor and compare with supervised and CLIP features.
- Discuss the limitations of VAEs and the link to modern generative models (diffusion, flow matching).

## Preview of the work

1. **Data setup.** unlabelled training images at low resolution (e.g. 64×64).
2. **Convolutional autoencoder.** encoder, bottleneck and decoder with transposed convolutions.
3. **Variational autoencoder.** loss, reparameterisation and β-weighting of the KL term.
4. **Generation and interpolation.** sampling from the prior and moving through the latent space.
5. **Latent-space analysis.** 2-D projection by class.
6. **Representation reuse.** linear classifier on latent codes.
7. **Conclusions.** what the generative model captures about garments.
