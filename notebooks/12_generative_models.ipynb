# 12 — Generative Models: Autoencoders and VAE

This notebook explores **generative models**, which learn the distribution of the training data and can sample new examples from it. A convolutional autoencoder and a **variational autoencoder (VAE)** are trained on H&M garment images to learn a compact latent representation without using the labels.

The latent space is analysed (structure by class, interpolation between garments, generation of new samples), and the learned encoder is reused as a feature extractor for classification, linking generative modelling with self-supervised representation learning.

## Generación frente a embeddings: alcance de la extensión

Interpretamos la referencia a GenAI como otra posible vía de aprendizaje, no como una obligación de construir un gran modelo generativo. Representar una prenda o una palabra mediante un vector no implica generar ejemplos nuevos. Un autoencoder básico aprende a reconstruir y comprimir; no define por sí solo un procedimiento justificado para muestrear imágenes nuevas. El VAE añade una distribución latente y un objetivo probabilístico que permiten estudiar muestreo e interpolación.

En una posible rama textual, **autorregresión** significa modelar una secuencia como `p(x₁,…,xₜ) = ∏ᵢ p(xᵢ | x₁,…,xᵢ₋₁)` y generar sucesivamente los siguientes tokens. Una RNN/LSTM/GRU con una cabeza de siguiente token o un Transformer causal pueden utilizar ese objetivo. La generación depende del objetivo y del acceso al contexto, no solo del nombre de la arquitectura.

BERT original es un encoder bidireccional con predicción de tokens enmascarados, no el ejemplo causal autorregresivo del apunte. Aquí lo usamos para explicar el contraste con GPT, sin afirmar que el profesor dijera que eran equivalentes. Las [fuentes de BERT](https://arxiv.org/abs/1810.04805) y [GPT-2](https://cdn.openai.com/better-language-models/language_models_are_unsupervised_multitask_learners.pdf) permiten verificar la distinción.

El alcance visual previsto sigue siendo un autoencoder y, si es viable, un VAE pequeño. La generación textual es solo una propuesta de ampliación. Se valorará lo aprendido sobre representación, reconstrucción, distribución y limitaciones, no el realismo máximo de las muestras. Las extensiones no sustituyen las arquitecturas básicas ni las ablaciones del núcleo del proyecto.


## Objectives

- Train a convolutional autoencoder and a VAE on low-resolution garment images.
- Explain the VAE objective: reconstruction term plus KL divergence and the reparameterisation trick.
- Visualise reconstructions, random samples and latent-space interpolations.
- Project the latent space in 2-D coloured by garment class.
- Use the encoder as a frozen feature extractor and compare with supervised and CLIP features.
- Discuss the limitations of VAEs and the link to modern generative models (diffusion, flow matching).

## Preview of the work

1. **Data setup.** unlabelled training images at low resolution (e.g. 64×64).
2. **Convolutional autoencoder.** encoder, bottleneck and decoder with transposed convolutions.
3. **Variational autoencoder.** loss, reparameterisation and β-weighting of the KL term.
4. **Generation and interpolation.** sampling from the prior and moving through the latent space.
5. **Latent-space analysis.** 2-D projection by class.
6. **Representation reuse.** linear classifier on latent codes.
7. **Conclusions.** what the generative model captures about garments.
